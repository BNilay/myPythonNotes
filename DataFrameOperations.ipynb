```python
# GROUP BY
```

```python
salary_dict = {"Programming Language":["Python","Python","Python","Java","Java","Swift"],"Name":["A","B","C","D","E","F"],
"Salary":[100,90,80,70,60,50]}
```

```python
salary_frame = pd.DataFrame(salary_dict)
salary_frame
```

```python
group_object = salary_frame.groupby("Programming Language")
```

```python
group_object["Salary"].mean()
```

```python
group_object["Salary"].min()
```

```python
group_object["Salary"].max()
```

```python
group_object.max("Salary")
```

```python
group_object.describe()
```

```python
my_dict1 = {"Name":["A","B","C","D"],
           "Sports":["Bas","foot","tennis","run"],
           "Calories":[100,200,300,400]}
```

```python
my_frame1 = pd.DataFrame(my_dict1)
my_frame1
```

```python
my_dict2 = {"Name":["E","F","G","H"],
           "Sports":["Bas","foot","tennis","run"],
           "Calories":[200,50,300,440]}
```

```python
my_frame2 = pd.DataFrame(my_dict2,index=[4,5,6,7])
my_frame2
```

```python
my_dict3 = {"Name":["I","J","K","L"],
           "Sports":["Bas","foot","tennis","run"],
           "Calories":[300,150,320,410]}
```

```python
my_frame3 = pd.DataFrame(my_dict3,index=[8,9,10,11])
my_frame3
```

```python
#concatenation
```

```python
# data frameler  birleştirilir
```

```python
pd.concat([my_frame1,my_frame2,my_frame3])
```

```python
#merge
```

```python
#aynı olan verileri kullanrak iç içe geçirme
```

```python
pd.merge(my_frame1,my_frame2,on="Sports")
```
